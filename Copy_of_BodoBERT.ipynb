```python
import pandas as pd

df = pd.read_excel('/content/pos_tagged.xlsx')
display(df.head())
```

```python
def parse_pos_tagged_string(text):
    words = []
    pos_tags = []
    if pd.isna(text):
        return [], []
    # Split the string by spaces to get individual word\tag units
    word_tag_units = text.strip().split(' ')
    for unit in word_tag_units:
        if unit:
            # Split each unit by the last backslash to separate word and tag
            parts = unit.rsplit('\\', 1)
            if len(parts) == 2:
                words.append(parts[0])
                pos_tags.append(parts[1])
            else:
                # If no backslash is found, treat the whole unit as a word and tag as None/empty
                words.append(unit)
                pos_tags.append(None) # Or consider 'UNKNOWN_TAG' based on expected data quality
    return words, pos_tags

# Apply the function to the 'Value' column to create new 'Words' and 'POS_Tags' columns
df['Words'], df['POS_Tags'] = zip(*df['Value'].apply(parse_pos_tagged_string))

# Display the DataFrame with the new columns
display(df.head())
```

```python
def tokenize_and_align_labels(sentences, labels):

    tokenized_inputs = tokenizer(
        sentences,
        truncation=True,
        padding=False,
        is_split_into_words=True
    )

    aligned_labels = []

    for i, label in enumerate(labels):

        word_ids = tokenized_inputs.word_ids(batch_index=i)

        previous_word_idx = None

        label_ids = []

        for word_idx in word_ids:

            if word_idx is None:
                label_ids.append(-100)

            elif word_idx != previous_word_idx:
                label_ids.append(label2id[label[word_idx]])

            else:
                label_ids.append(-100)

            previous_word_idx = word_idx

        aligned_labels.append(label_ids)

    tokenized_inputs["labels"] = aligned_labels

    return tokenized_inputs
```

```python
train_dataset = Dataset.from_dict({
    "tokens": train_sentences,
    "labels": train_labels
})

val_dataset = Dataset.from_dict({
    "tokens": val_sentences,
    "labels": val_labels
})

test_dataset = Dataset.from_dict({
    "tokens": test_sentences,
    "labels": test_labels
})
```

```python
train_dataset = train_dataset.map(
    lambda x: tokenize_and_align_labels(
        x["tokens"],
        x["labels"]
    ),
    batched=True
)

val_dataset = val_dataset.map(
    lambda x: tokenize_and_align_labels(
        x["tokens"],
        x["labels"]
    ),
    batched=True
)

test_dataset = test_dataset.map(
    lambda x: tokenize_and_align_labels(
        x["tokens"],
        x["labels"]
    ),
    batched=True
)
```

```python
import numpy as np
import evaluate

seqeval = evaluate.load("seqeval")
```

```python
def compute_metrics(eval_pred):

    predictions, labels = eval_pred

    predictions = np.argmax(predictions, axis=2)

    true_predictions = [
        [id2label[p] for (p, l) in zip(prediction, label) if l != -100]
        for prediction, label in zip(predictions, labels)
    ]

    true_labels = [
        [id2label[l] for (p, l) in zip(prediction, label) if l != -100]
        for prediction, label in zip(predictions, labels)
    ]

    results = seqeval.compute(
        predictions=true_predictions,
        references=true_labels
    )

    return {
        "precision": results["overall_precision"],
        "recall": results["overall_recall"],
        "f1": results["overall_f1"],
        "accuracy": results["overall_accuracy"],
    }
```

```python
from transformers import TrainingArguments

training_args = TrainingArguments(

    output_dir="./bodo_pos_model",

    eval_strategy="epoch",

    save_strategy="epoch",

    learning_rate=5e-5, # Increased learning rate for hyperparameter tuning

    per_device_train_batch_size=8,

    per_device_eval_batch_size=8,

    num_train_epochs=10,

    weight_decay=0.01,

    logging_steps=50,

    load_best_model_at_end=True,

    metric_for_best_model="f1",

    greater_is_better=True,

    save_total_limit=2,

    report_to="none",

    # Updated from warmup_ratio to warmup_steps as recommended by the warning.
    # Calculated as 10% of total steps: 0.1 * (num_train_epochs * len(train_dataset) / per_device_train_batch_size)
    # 0.1 * (10 * 4804 / 8) = 0.1 * (10 * 600.5) = 600.5, rounded to 601.
    warmup_steps=601
)
```

```python
from collections import Counter
import torch

# Flatten the list of lists of true labels to count occurrences
all_train_labels = [label for sublist in train_labels for label in sublist]

# Count the occurrences of each label
label_counts = Counter(all_train_labels)

# Calculate inverse class weights
# The idea is to give less frequent classes higher weights
total_labels = sum(label_counts.values())
max_count = max(label_counts.values())

class_weights = torch.zeros(len(label2id))

for label_name, count in label_counts.items():
    if label_name in label2id:
        idx = label2id[label_name]
        # Inverse frequency weighting or inverse square root frequency
        # A common approach: 1 / frequency or N / (N_k * K) where N is total samples, N_k is samples for class k, K is number of classes
        # Here, I'll use a simpler inverse frequency scaled by total_labels/max_count to keep values manageable
        class_weights[idx] = total_labels / (count * len(label2id))

# Normalize weights so their sum is 1, or scale them as needed. Sometimes just raw inverse is used.
# For now, let's ensure minimum weight is not zero if a class isn't present (though all unique tags should be)
# and potentially normalize them.

# If a label doesn't appear in the training data, its count will be 0, leading to division by zero.
# We should handle cases where a label might not appear in all_train_labels if unique_tags contains more tags than present in train_labels.
# Based on previous checks, all unique_tags should be in train_labels, but it's good practice.
# Set weight to a small value for classes not appearing in train_labels to avoid inf, or to 0 depending on desired behavior.
for idx, label_name in id2label.items():
    if label_counts[label_name] == 0:
        class_weights[idx] = 0.0 # Or a very small number like 0.01

print("Class Weights:")
for idx, weight in enumerate(class_weights):
    print(f"  {id2label[idx]}: {weight:.4f}")
```

```python
from torch.nn import CrossEntropyLoss

class CustomTrainer(Trainer):
    def compute_loss(self, model, inputs, return_outputs=False, num_items_in_batch=None, **kwargs):
        labels = inputs.pop("labels")
        # forward pass
        outputs = model(**inputs)
        logits = outputs.get("logits")
        # compute custom loss
        loss_fct = CrossEntropyLoss(weight=class_weights.to(model.device))
        loss = loss_fct(logits.view(-1, self.model.config.num_labels), labels.view(-1))
        return (loss, outputs) if return_outputs else loss
```

```python
# Re-initialize the model to ensure it's in a fresh state before re-training with new parameters
model = AutoModelForTokenClassification.from_pretrained(
    MODEL_NAME,
    num_labels=len(label2id),
    id2label=id2label,
    label2id=label2id
)

# Instantiate the custom trainer
trainer = CustomTrainer(
    model=model,
    args=training_args,
    train_dataset=train_dataset,
    eval_dataset=val_dataset,
    processing_class=tokenizer,
    data_collator=data_collator,
    compute_metrics=compute_metrics,
)

# Train the model with the updated learning rate and weighted loss
trainer.train()
```

```python
test_results = trainer.evaluate(test_dataset)

print(f"New Test Accuracy: {test_results['eval_accuracy']:.4f}")
print(f"New Test F1 Score: {test_results['eval_f1']:.4f}")
```
